```python
import os
import cv2
import mediapipe as mp


OUTPUT_FOLDER =r"C:\Users\vutuk\OneDrive\Desktop\FaceRecognitionProject\Cropped_face"

os.makedirs(OUTPUT_FOLDER,exist_ok=True)

# FACE DETECTION + CROPPING + PREPROCESSING
def process_image(image_path,save_path):
    image = cv2.imread(image_path)

    if image is None:
        return False

    rgb = cv2.cvtColor(image,cv2.COLOR_BGR2RGB)

    mp_image = mp.Image(image_format=mp.ImageFormat.SRGB,data=rgb)

    result = detector.detect(mp_image)

    # No face found
    if not result.face_landmarks:
        return False

    face = result.face_landmarks[0]

    h,w,_ = image.shape

    # Bounding Box

    xs = [lm.x * w for lm in face]

    ys = [lm.y * h for lm in face]

    x_min = int(min(xs))
    x_max = int(max(xs))

    y_min = int(min(ys))
    y_max = int(max(ys))
    
    # Add Padding

    padding = 30

    x_min = max(0,x_min-padding)

    y_min = max(0,y_min-padding)

    x_max = min(w,x_max+padding)

    y_max = min(h,y_max+padding)
    
    # Crop Face
    
    face_crop = image[y_min:y_max,x_min:x_max]

    if face_crop.size == 0:
        return False
        
    # PREPROCESSIN

    face_crop = cv2.resize(face_crop,(224,224))

    face_crop = cv2.GaussianBlur(face_crop,(3,3),0)

    # SAVE

    cv2.imwrite(save_path,face_crop)

    return True
```

```python
# apply on data set
from tqdm import tqdm

DATASET_PATH =r'C:\Users\vutuk\OneDrive\Desktop\FaceRecognitionProject\datasets'

processed_count = 0

failed_count = 0

for person in os.listdir(DATASET_PATH):

    person_folder = os.path.join(DATASET_PATH,person)

    if not os.path.isdir(person_folder):
        continue

    save_person_folder = os.path.join(OUTPUT_FOLDER,person)

    os.makedirs(save_person_folder,exist_ok=True)

    print(f"\nProcessing {person}")

    for image_name in tqdm(os.listdir(person_folder)):

        image_path = os.path.join(person_folder,image_name)

        save_path = os.path.join(save_person_folder,image_name)

        success = process_image(image_path,save_path)

        if success:

            processed_count += 1

        else:

            failed_count += 1
```

```python
import cv2
import joblib
import pandas as pd
import mediapipe as mp

from keras_facenet import FaceNet
from mediapipe.tasks import python
from mediapipe.tasks.python import vision


# =====================================
# LOAD MODEL
# =====================================

model_data = joblib.load(r"C:\Users\vutuk\OneDrive\Desktop\FaceRecognitionProject\models\face_recognition_facenet_model.pkl"
)

knn = model_data["knn"]
encoder = model_data["encoder"]
scaler = model_data["scaler"]

print("Model Loaded")


# =====================================
# LOAD STUDENT DATA
# =====================================

students = pd.read_csv(r"C:\Users\vutuk\OneDrive\Desktop\FaceRecognitionProject\metadata\students.csv"
)

print("Metadata Loaded")


# =====================================
# LOAD FACENET
# =====================================

embedder = FaceNet()

print("FaceNet Loaded")


# =====================================
# LOAD MEDIAPIPE DETECTOR
# =====================================

MODEL_PATH =r"C:\Users\vutuk\OneDrive\Desktop\FaceRecognitionProject\face_landmarker.task"

##Initialize Face Landmarker

base_options = python.BaseOptions(
    model_asset_path=MODEL_PATH
)

options = vision.FaceLandmarkerOptions(
    base_options=base_options,
    num_faces=1
)

detector = vision.FaceLandmarker.create_from_options(
    options
)


print("Detector Loaded")


# =====================================
# FACE DETECTION FUNCTION
# =====================================

def detect_and_crop_face(frame):

    rgb = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2RGB
    )

    mp_image = mp.Image(
        image_format=mp.ImageFormat.SRGB,
        data=rgb
    )

    result = detector.detect(
        mp_image
    )

    if not result.face_landmarks:
        return None

    face = result.face_landmarks[0]

    h, w, _ = frame.shape

    xs = [lm.x * w for lm in face]
    ys = [lm.y * h for lm in face]

    x_min = int(min(xs))
    x_max = int(max(xs))

    y_min = int(min(ys))
    y_max = int(max(ys))

    padding = 30

    x_min = max(0, x_min-padding)
    y_min = max(0, y_min-padding)

    x_max = min(w, x_max+padding)
    y_max = min(h, y_max+padding)

    face_crop = frame[
        y_min:y_max,
        x_min:x_max
    ]

    if face_crop.size == 0:
        return None

    face_crop = cv2.resize(
        face_crop,
        (160,160)
    )

    return face_crop


# =====================================
# GET STUDENT INFO
# =====================================

def get_student_info(name):

    row = students[
        students["Name"] == name
    ]

    if len(row) == 0:
        return None

    return row.iloc[0].to_dict()


# =====================================
# START CAMERA
# =====================================

cap = cv2.VideoCapture(0)

while True:

    success, frame = cap.read()

    if not success:
        break

    face_crop = detect_and_crop_face(
        frame
    )

    if face_crop is not None:

        # FaceNet Embedding

        embedding = embedder.embeddings(
            [face_crop]
        )[0]

        embedding = embedding.reshape(
            1,-1
        )

        embedding = scaler.transform(
            embedding
        )

        prediction = knn.predict(
            embedding
        )

        name = encoder.inverse_transform(
            prediction
        )[0]

        info = get_student_info(
            name
        )

        print("\n====================")

        if info:

            for key, value in info.items():

                print(
                    f"{key}: {value}"
                )

        print("====================")

    cv2.imshow(
        "Student Recognition",
        frame
    )

    if cv2.waitKey(1) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
